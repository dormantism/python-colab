```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

np.random.seed(0)

#x와y 데이터 생성
x=2*np.random.rand(100,1)
y=4+3*x+np.random.randn(100,1)

#산점도 출력
plt.scatter(x,y)
plt.title("산점도")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

```python
#선형 회귀 모델 적합
lin_reg=LinearRegression()
lin_reg.fit(x,y)

#회귀 계수 출력
print("절편:",lin_reg.intercept_)
print("기울기:",lin_reg.coef_)

#회귀선 시각화
x_new=np.array([[0],[2]])
y_pred=lin_reg.predict(x_new)

plt.plot(x_new, y_pred, "r-", label="회귀선")
plt.scatter(x,y,alpha=0.7)
plt.xlabel('x')
plt.ylabel('y')
plt.title("선형회귀 결과")
plt.legend()
plt.show()
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

#예제 데이터 생성
x=np.array([[1],[2],[3],[4],[5]])
y=np.array([[2],[4],[5],[4],[5]])

#모델 생성 및 학습
model=LinearRegression()
model.fit(x,y)

#예측값 계산
y_pred=model.predict(x)

#R-squard 계산
r2=r2_score(y,y_pred)

#MSE 계산
mse=mean_squared_error(y,y_pred)

#회귀 계수 출력
print("절편 (Intercept):",model.intercept_[0])
print("기울기 (Slope):",model.coef_[0][0])
print("결정계수 (R^2):",r2)
print("평균제곱오차 (MSE):",mse)
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
import matplotlib.pyplot as plt

#CSV 파일 불러오기
df=pd.read_csv("/content/sample_data/housing_data.csv")


#입력 변수(면적)와 출력 변수(가격) 분리
X=df[['Area (sq m)']].values #2D array
y=df[['Price ($)']].values #2D array

#선형 회기 모델 생성 및 학습
model=LinearRegression()
model.fit(X,y)

#회귀 계수 및 절편 출력
intercept=model.intercept_[0]
slope=model.coef_[0][0]
print(f"절편 (Intercept): {intercept:.2f}")
print(f"기울기 (Slope): {slope:.2f}")
```

```python
#예측 및 성능 평가
y_pred=model.predict(X)
r2=r2_score(y,y_pred)
mse=mean_squared_error(y,y_pred)

print(f"결정 계수(R:2): {r2:.3f}")
print(f"평균 제곱 오차(MSE): {mse:.2f}")

#시각화
plt.scatter(X,y,color='blue',alpha=0.6, label='Real data')
plt.plot(X,y_pred,color='red',label='Regression line')
plt.xlabel("Area ($m^2$)")
plt.ylabel("Price ($)'")
plt.title("Housing area vs Price regression analysis")
plt.legend()
plt.grid(True)
plt.show()
```
