```python
import numpy as np
import scipy as sp
import scipy.stats
import matplotlib.pyplot as plt
import seaborn as sns

N_1=10
mu_1=0
sigma_1=1
N_2=10
mu_2=0.5
sigma_2=1
np.random.seed(0)
x1=sp.stats.norm(mu_1,sigma_1).rvs(N_1)
x2=sp.stats.norm(mu_2,sigma_2).rvs(N_2)
ax=sns.distplot(x1, kde=False, fit=sp.stats.norm,label="Data set of No.1")
ax=sns.distplot(x2, kde=False, fit=sp.stats.norm,label="Data set of No.2")
ax.lines[0].set_linestyle(":")
plt.legend()
plt.show
```
